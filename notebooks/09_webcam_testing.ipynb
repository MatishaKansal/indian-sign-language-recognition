# 09 - Local webcam testing (15 classes)

Choose the compact landmark-only baseline or improved landmark + MobileNet model. Run notebooks 02 through 07 to create the 15-class checkpoints. Notebook 08 reports image-test accuracy; webcam performance on new signers has not been measured.

```python
import sys
from pathlib import Path
PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name.lower() == 'notebooks':
    PROJECT_ROOT = PROJECT_ROOT.parent
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))
from src.config import load_config, project_path
from src.ensemble_inference import load_landmark_predictor, run_landmark_webcam

MODEL = 'improved'  # 'compact_baseline' or 'improved'
CAMERA_INDEX = 0
CONFIDENCE_THRESHOLD = 0.50
checkpoint_dir = project_path(load_config(), 'checkpoints')
paths = {
    'compact_baseline': checkpoint_dir / 'landmark_mlp_best.pt',
    'improved': checkpoint_dir / 'robust_landmark_ensemble_best.pt',
}
if MODEL not in paths:
    raise ValueError('MODEL must be compact_baseline or improved.')
checkpoint_path = paths[MODEL]
predictor = load_landmark_predictor(checkpoint_path)
expected_type = {'compact_baseline': 'landmark_mlp', 'improved': 'robust_landmark_ensemble'}[MODEL]
if predictor.model_type != expected_type:
    raise ValueError(f'Unexpected checkpoint model type: {predictor.model_type}')
print(f'Loaded {MODEL}: {len(predictor.class_to_idx)} classes. Run the next cell to start the camera.')
```

```python
run_landmark_webcam(checkpoint_path, camera_index=CAMERA_INDEX, confidence_threshold=CONFIDENCE_THRESHOLD)
```
